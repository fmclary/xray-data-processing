```python
import numpy as np
import os
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def select_ldos(element):
    dats = []
    for file in os.listdir('feff_outputs/'):
        if 'ldos' in file:
            if element in file:
                dats.append('feff_outputs/'+file)
    return dats
```

```python
def load_ldos(filename):
    data = pd.read_csv(filename, sep='\s+', header=None, names=['energy', 'sDOS', 'pDOS', 'dDOS', 'fDOS'], comment = '#').astype(float)
    return data
```

```python
def get_index(list, material, absorber):
    filename = 'feff_outputs/' + 'ldos00_' + material + '_' + absorber + '.dat'
    ix = list.index(filename)
    return ix
```

```python
dats = select_ldos('Mn3O4')
```

```python
dats
```

```python
ldos_df = []
for file in dats:
    df = load_ldos(file)
    ldos_df.append(df)
```

```python
ell = ['s', 'p', 'd']
for l in ell:
    plt.plot(ldos_df[1]['energy'], ldos_df[1][l + 'DOS'], label = l + 'DOS')

plt.xlabel('Energy')
plt.ylabel('DOS')
plt.title('Tetrahedral Mn2+ density of states in Mn3O4')
plt.legend()
plt.show()
```

```python
ell = ['s', 'p', 'd']
for l in ell:
    plt.plot(ldos_df[0]['energy'], ldos_df[0][l + 'DOS'], label = l + 'DOS')

plt.xlabel('Energy')
plt.ylabel('DOS')
plt.title('Octahedral Mn3+ density of states in Mn3O4')
plt.legend()
plt.show()
```

```python
dats = select_ldos('Fe3O4')

ldos_df = []
for file in dats:
    df = load_ldos(file)
    ldos_df.append(df)

    ell = ['s', 'p', 'd']
for l in ell:
    plt.plot(ldos_df[get_index(dats, 'Fe3O4', 'Fe1')]['energy'], ldos_df[get_index(dats, 'Fe3O4', 'Fe1')][l + 'DOS'], label = l + 'DOS')

plt.xlabel('Energy')
plt.ylabel('DOS')
plt.title('Octahedral Fe2+ density of states in Fe3O4')
plt.legend()
plt.show()
```
