```python
import numpy as np
import os
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def load_spec_espace(filename):
    data = np.loadtxt(filename, comments = "#").T
    data = np.asarray([data[1], data[3]])
    data = np.transpose(data)
    data = pd.DataFrame(data)
    return data

def load_spec_kspace(filename):
    data = np.loadtxt(filename, comments = "#").T
    data = np.asarray([data[2], data[5]])
    data = np.transpose(data)
    data = pd.DataFrame(data)
    return data
```

```python
dats=[]
for file in os.listdir():
    if 'chlorite' in file:
        dats.append(file)

FeM1_xanes=[]
FeM1_xes=[]
FeM1_exafs=[]
FeM2_xanes=[]
FeM2_xes=[]
FeM2_exafs=[]
FeM3_xanes=[]
FeM3_xes=[]
FeM3_exafs=[]
FeM4_xanes=[]
FeM4_xes=[]
FeM4_exafs=[]
for file in dats:
    if 'fem1' in file:
        if 'xanes' in file:
            FeM1_xanes.append(file)
        if 'xes' in file:
            FeM1_xes.append(file)
        if 'exafs' in file:
            FeM1_exafs.append(file)
    if 'fem2' in file:
        if 'xanes' in file:
            FeM2_xanes.append(file)
        if 'xes' in file:
            FeM2_xes.append(file)
        if 'exafs' in file:
            FeM2_exafs.append(file)
    if 'fem3' in file:
        if 'xanes' in file:
            FeM3_xanes.append(file)
        if 'xes' in file:
            FeM3_xes.append(file)
        if 'exafs' in file:
            FeM3_exafs.append(file)
    if 'fem4' in file:
        if 'xanes' in file:
            FeM4_xanes.append(file)
        if 'xes' in file:
            FeM4_xes.append(file)
        if 'exafs' in file:
            FeM4_exafs.append(file)
```

```python
FeM1_xanes_df = []
FeM1_xes_df = []
FeM1_exafs_e_df = []
FeM1_exafs_k_df = []
for file in FeM1_xanes:
    df = load_spec_espace(file)
    FeM1_xanes_df.append(df)
for file in FeM1_xes:
    df = load_spec_espace(file)
    FeM1_xes_df.append(df)
for file in FeM1_exafs:
    df1 = load_spec_espace(file)
    FeM1_exafs_e_df.append(df1)
    df2 = load_spec_kspace(file)
    FeM1_exafs_k_df.append(df2)

FeM2_xanes_df = []
FeM2_xes_df = []
FeM2_exafs_e_df = []
FeM2_exafs_k_df = []
for file in FeM2_xanes:
    df = load_spec_espace(file)
    FeM2_xanes_df.append(df)
for file in FeM2_xes:
    df = load_spec_espace(file)
    FeM2_xes_df.append(df)
for file in FeM2_exafs:
    df1 = load_spec_espace(file)
    FeM2_exafs_e_df.append(df1)
    df2 = load_spec_kspace(file)
    FeM2_exafs_k_df.append(df2)

FeM3_xanes_df = []
FeM3_xes_df = []
FeM3_exafs_e_df = []
FeM3_exafs_k_df = []
for file in FeM3_xanes:
    df = load_spec_espace(file)
    FeM3_xanes_df.append(df)
for file in FeM3_xes:
    df = load_spec_espace(file)
    FeM3_xes_df.append(df)
for file in FeM3_exafs:
    df1 = load_spec_espace(file)
    FeM3_exafs_e_df.append(df1)
    df2 = load_spec_kspace(file)
    FeM3_exafs_k_df.append(df2)

FeM4_xanes_df = []
FeM4_xes_df = []
FeM4_exafs_e_df = []
FeM4_exafs_k_df = []
for file in FeM4_xanes:
    df = load_spec_espace(file)
    FeM4_xanes_df.append(df)
for file in FeM4_xes:
    df = load_spec_espace(file)
    FeM4_xes_df.append(df)
for file in FeM4_exafs:
    df1 = load_spec_espace(file)
    FeM4_exafs_e_df.append(df1)
    df2 = load_spec_kspace(file)
    FeM4_exafs_k_df.append(df2)
```

```python
FeM1_xanes_avg=pd.concat([FeM1_xanes_df[0][0],FeM1_xanes_df[0][1],FeM1_xanes_df[1][1],FeM1_xanes_df[2][1]], axis=1)
FeM1_xanes_avg['avg']=FeM1_xanes_avg.iloc[:,[1,2,3]].mean(axis=1)
FeM2_xanes_avg=pd.concat([FeM2_xanes_df[0][0],FeM2_xanes_df[0][1],FeM2_xanes_df[1][1],FeM2_xanes_df[2][1]], axis=1)
FeM2_xanes_avg['avg']=FeM2_xanes_avg.iloc[:,[1,2,3]].mean(axis=1)
FeM3_xanes_avg=pd.concat([FeM3_xanes_df[0][0],FeM3_xanes_df[0][1],FeM3_xanes_df[1][1],FeM3_xanes_df[2][1]], axis=1)
FeM3_xanes_avg['avg']=FeM3_xanes_avg.iloc[:,[1,2,3]].mean(axis=1)
FeM4_xanes_avg=pd.concat([FeM4_xanes_df[0][0],FeM4_xanes_df[0][1],FeM4_xanes_df[1][1],FeM4_xanes_df[2][1]], axis=1)
FeM4_xanes_avg['avg']=FeM4_xanes_avg.iloc[:,[1,2,3]].mean(axis=1)
```

```python
FeM1_xes_avg=pd.concat([FeM1_xes_df[0][0],FeM1_xes_df[0][1],FeM1_xes_df[1][1],FeM1_xes_df[2][1]], axis=1)
FeM1_xes_avg['avg']=FeM1_xes_avg.iloc[:,[1,2,3]].mean(axis=1)
FeM2_xes_avg=pd.concat([FeM2_xes_df[0][0],FeM2_xes_df[0][1],FeM2_xes_df[1][1],FeM2_xes_df[2][1]], axis=1)
FeM2_xes_avg['avg']=FeM2_xes_avg.iloc[:,[1,2,3]].mean(axis=1)
FeM3_xes_avg=pd.concat([FeM3_xes_df[0][0],FeM3_xes_df[0][1],FeM3_xes_df[1][1],FeM3_xes_df[2][1]], axis=1)
FeM3_xes_avg['avg']=FeM3_xes_avg.iloc[:,[1,2,3]].mean(axis=1)
FeM4_xes_avg=pd.concat([FeM4_xes_df[0][0],FeM4_xes_df[0][1],FeM4_xes_df[1][1],FeM4_xes_df[2][1]], axis=1)
FeM4_xes_avg['avg']=FeM4_xes_avg.iloc[:,[1,2,3]].mean(axis=1)
```

```python
FeM1_exafs_e_avg=pd.concat([FeM1_exafs_e_df[0][0],FeM1_exafs_e_df[0][1],FeM1_exafs_e_df[1][1],FeM1_exafs_e_df[2][1]], axis=1)
FeM1_exafs_e_avg['avg']=FeM1_exafs_e_avg.iloc[:,[1,2,3]].mean(axis=1)
FeM2_exafs_e_avg=pd.concat([FeM2_exafs_e_df[0][0],FeM2_exafs_e_df[0][1],FeM2_exafs_e_df[1][1],FeM2_exafs_e_df[2][1]], axis=1)
FeM2_exafs_e_avg['avg']=FeM2_exafs_e_avg.iloc[:,[1,2,3]].mean(axis=1)
FeM3_exafs_e_avg=pd.concat([FeM3_exafs_e_df[0][0],FeM3_exafs_e_df[0][1],FeM3_exafs_e_df[1][1],FeM3_exafs_e_df[2][1]], axis=1)
FeM3_exafs_e_avg['avg']=FeM3_exafs_e_avg.iloc[:,[1,2,3]].mean(axis=1)
FeM4_exafs_e_avg=pd.concat([FeM4_exafs_e_df[0][0],FeM4_exafs_e_df[0][1],FeM4_exafs_e_df[1][1],FeM4_exafs_e_df[2][1]], axis=1)
FeM4_exafs_e_avg['avg']=FeM4_exafs_e_avg.iloc[:,[1,2,3]].mean(axis=1)
```

```python
FeM1_exafs_k_avg=pd.concat([FeM1_exafs_k_df[0][0],FeM1_exafs_k_df[0][1],FeM1_exafs_k_df[1][1],FeM1_exafs_k_df[2][1]], axis=1)
FeM1_exafs_k_avg['avg']=FeM1_exafs_k_avg.iloc[:,[1,2,3]].mean(axis=1)
FeM2_exafs_k_avg=pd.concat([FeM2_exafs_k_df[0][0],FeM2_exafs_k_df[0][1],FeM2_exafs_k_df[1][1],FeM2_exafs_k_df[2][1]], axis=1)
FeM2_exafs_k_avg['avg']=FeM2_exafs_k_avg.iloc[:,[1,2,3]].mean(axis=1)
FeM3_exafs_k_avg=pd.concat([FeM3_exafs_k_df[0][0],FeM3_exafs_k_df[0][1],FeM3_exafs_k_df[1][1],FeM3_exafs_k_df[2][1]], axis=1)
FeM3_exafs_k_avg['avg']=FeM3_exafs_k_avg.iloc[:,[1,2,3]].mean(axis=1)
FeM4_exafs_k_avg=pd.concat([FeM4_exafs_k_df[0][0],FeM4_exafs_k_df[0][1],FeM4_exafs_k_df[1][1],FeM4_exafs_k_df[2][1]], axis=1)
FeM4_exafs_k_avg['avg']=FeM4_exafs_k_avg.iloc[:,[1,2,3]].mean(axis=1)
```

```python

plt.plot(FeM1_xanes_avg[0],FeM1_xanes_avg['avg'], label = 'FeM1')
plt.plot(FeM2_xanes_avg[0],FeM2_xanes_avg['avg'], label = 'FeM2')
plt.plot(FeM3_xanes_avg[0],FeM3_xanes_avg['avg'], label = 'FeM3')
plt.plot(FeM4_xanes_avg[0],FeM4_xanes_avg['avg'], label = 'FeM4')
plt.xlabel('Energy')
plt.ylabel('μ')
plt.title('Chlorite XANES')
plt.legend()
plt.show()
```
