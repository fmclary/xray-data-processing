```python
import numpy as np
import os
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def load_spec_espace(filename):
    data = np.loadtxt(filename, comments = "#").T
    data = np.asarray([data[1], data[3]])
    data = np.transpose(data)
    data = pd.DataFrame(data)
    return data

def load_spec_kspace(filename):
    data = np.loadtxt(filename, comments = "#").T
    data = np.asarray([data[2], data[5]])
    data = np.transpose(data)
    data = pd.DataFrame(data)
    return data
```

```python
# Make lists for each spec type

dats = []
for file in os.listdir('feff_outputs/'):
    if 'xmu_MnO2' in file:
        dats.append('feff_outputs/'+file)

xanes = []
xes = []
exafs = []
for file in dats:
    if 'xanes' in file:
        xanes.append(file)
    if 'xes' in file:
        xes.append(file)
    if 'exafs' in file:
        exafs.append(file)
```

```python
xanes
```

```python
# Convert lists of files to lists of dataframes

xanes_df = []
xes_df = []
exafs_e_df = []
exafs_k_df = []
for file in xanes:
    df = load_spec_espace(file)
    xanes_df.append(df)
for file in xes:
    df = load_spec_espace(file)
    xes_df.append(df)
for file in exafs:
    df1 = load_spec_espace(file)
    exafs_e_df.append(df1)
    df2 = load_spec_kspace(file)
    exafs_k_df.append(df2)
```

```python
plt.plot(xanes_df[1][0], xanes_df[1][1], label = 'Mn0')
plt.plot(xanes_df[0][0], xanes_df[0][1], label = 'Mn2')
plt.legend()
plt.show()
```

```python
plt.plot(xes_df[1][0], xes_df[1][1], label = 'Mn0')
plt.plot(xes_df[0][0], xes_df[0][1], label = 'Mn2')
plt.legend()
plt.show()
```

```python
plt.plot(exafs_k_df[1][0], exafs_k_df[1][1], label = 'Mn0')
plt.plot(exafs_k_df[0][0], exafs_k_df[0][1], label = 'Mn2')
plt.legend()
plt.show()
```

```python
# Concatenate each df and create a new column that averages the y value

xanes_avg = pd.concat([xanes_df[0][0], xanes_df[0][1], xanes_df[1][1]], axis = 1)
xanes_avg['avg'] = xanes_avg.iloc[:,[1,2]].mean(axis = 1)
xes_avg = pd.concat([xes_df[0][0], xes_df[0][1], xes_df[1][1]], axis = 1)
xes_avg['avg'] = xes_avg.iloc[:,[1,2]].mean(axis = 1)
exafs_k_avg = pd.concat([exafs_k_df[0][0], exafs_k_df[0][1], exafs_k_df[1][1]], axis = 1)
exafs_k_avg['avg'] = exafs_k_avg.iloc[:,[1,2]].mean(axis = 1)
```

```python
xanes_avg.to_csv('feff_outputs/averaged_outputs/xmu_MnO2_xanes.dat', sep = '\t', columns = None, header = False)
xes_avg.to_csv('feff_outputs/averaged_outputs/xmu_MnO2_xes.dat', sep = '\t', columns = None, header = False)
exafs_k_avg.to_csv('feff_outputs/averaged_outputs/xmu_MnO2_exafs_k.dat', sep = '\t', columns = None, header = False)
```
